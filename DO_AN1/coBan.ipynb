```python
import cv2 
print(cv2.__version__)
```

```python
img = cv2.imread("chup1.jpg")

# Giảm kích thước ảnh gốc xuống 50%
img = cv2.resize(img, None, fx=0.5, fy=0.5) 

# Chuyển sang ảnh xám
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # làm xám

gauss = cv2.GaussianBlur(gray, (3,3), 0)  #làm mịn - giảm nhiễu

# edged = cv2.Canny(gauss, 50, 150)



# clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))   #tăng tương phản
# enhanced = clahe.apply(gauss)

thresh = cv2.adaptiveThreshold(gauss, 255,
                               cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                               cv2.THRESH_BINARY_INV, 11, 2)  #nhị phân hóa.


# kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (1,1))  #loại bỏ nhiễu nhỏ
# cleaned = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)  





# Hiển thị kết quả



cv2.imshow('Anh Xam',thresh)



cv2.waitKey(0)
cv2.destroyAllWindows()


```

```python

# 1. tìm tất cả contour (chỉ contour ngoài cùng)
cnts, hierarchy = cv2.findContours(thresh.copy(), 
                                   cv2.RETR_EXTERNAL, 
                                   cv2.CHAIN_APPROX_SIMPLE)

# 2. danh sách để chứa contour có khả năng là mốc
candidate_squares = []

# 3. duyệt qua từng contour
for c in cnts:
    area = cv2.contourArea(c)              # diện tích contour
    if area < 500:                         # loại bỏ contour quá nhỏ (nhiễu)
        continue

    x, y, w, h = cv2.boundingRect(c)      # lấy bounding box vuông chữ nhật
    ratio = w / float(h)

    # 4. kiểm tra tỉ lệ gần vuông và diện tích vừa phải
    
    if 0.8 < ratio < 1.2 and 500 < area < 5000:
        # 5. kiểm tra độ đặc (fill ratio)
        rect_area = w * h
        fill_ratio = area / float(rect_area)
        if fill_ratio > 0.75:
            candidate_squares.append(c)

output = img.copy()

for c in candidate_squares:
    cv2.drawContours(output, c, -1, (0, 255, 0), 2)  # Màu xanh, độ dày 2 px

cv2.imshow("Contours", output)
cv2.waitKey(0)
cv2.destroyAllWindows()

```
